#Object Oriented Way

MatplotLib allows the aspect ratio,DPI 

and figure size to be specified when the Figure object is created.

```python
fig = plt.figure(figsize=(20,8),dpi = 120)
axis1 = fig.add_axes([0.1,0.1,0.8,1])
#(%left,%bottom,%width,%height)
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.3,0.5,0.8])
axis2.plot(x,y)
```
